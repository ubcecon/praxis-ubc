# RISE slideshow setup

Run this notebook once. It does two things: it installs the RISE slideshow tool, and it
adjusts one JupyterLab setting so the slideshow loads correctly on JupyterOpen. Both changes
go in your home folder, so they stay set every time you log in.

Steps:

1. Run all the cells (Run menu, then Run All Cells).
2. Check that the last output cell says everything is set.
3. Restart your server using the instructions in the last cell. A browser refresh does not
   do this.

After that, open a notebook like docs/SOCI-280/soci_280_bert.ipynb, open the command palette with Ctrl+Shift+C, and run "Open as Reveal Slideshow". The slides open in a new tab.

```python
import sys, subprocess

# Install RISE into ~/.local so it stays installed after a server restart.
subprocess.run([sys.executable, "-m", "pip", "install", "--user",
                "--force-reinstall", "--no-deps", "jupyterlab_rise==0.43.1"], check=True)

# Enable the server extension. It may print a validation warning, that is fine, the config
# still gets written.
r = subprocess.run([sys.executable, "-m", "jupyter", "server", "extension",
                    "enable", "jupyterlab_rise", "--user"],
                   capture_output=True, text=True)
print(r.stdout, r.stderr)
print("Install step done. Run the next cell.")
```

```python
import json, pathlib

# JupyterOpen turns off the notebook cell-executor extension because it runs cells a
# different way. The RISE slideshow page still needs that extension, so without this the
# slideshow opens as a blank white page. This turns it back on for your account only, and
# writes to your home folder so it stays set after a restart.
p = pathlib.Path.home() / ".jupyter" / "labconfig" / "page_config.json"
p.parent.mkdir(parents=True, exist_ok=True)
cfg = json.loads(p.read_text()) if p.exists() else {}
cfg.setdefault("disabledExtensions", {})["@jupyterlab/notebook-extension:cell-executor"] = False
p.write_text(json.dumps(cfg, indent=2))
print("Wrote", p)
print("Done. Run the next cell, then restart your server.")
```

```python
import site, os, json, pathlib

ul = site.getusersitepackages()
print("user site enabled:", site.ENABLE_USER_SITE)
print("rise installed in home:", os.path.isdir(os.path.join(ul, "jupyterlab_rise")))

p = pathlib.Path.home() / ".jupyter" / "labconfig" / "page_config.json"
cfg = json.loads(p.read_text()) if p.exists() else {}
executor = cfg.get("disabledExtensions", {}).get("@jupyterlab/notebook-extension:cell-executor")
print("slideshow cell-executor set to load:", executor is False)
print("If all three say True, restart your server (see the last cell).")
```

## Restart your server

The server has to restart before it picks up these changes. A browser refresh will not work.

1. In the menu bar, go to File, then Hub Control Panel. You can also open
   https://open.jupyter.ubc.ca/jupyter/hub/home in a new tab.
2. Click Stop My Server and wait for it to stop.
3. Click Start My Server.
4. When JupyterLab loads again, open a notebook like docs/SOCI-280/soci_280_bert.ipynb.
   Open the command palette with Ctrl+Shift+C, type "Open as Reveal Slideshow", and run it.
   The slides open in a new tab.

Use "Open as Reveal Slideshow" from the command palette. Do not use "Render as Reveal Slideshow" from the View menu, that one does not work on JupyterOpen.

You do not need to run this setup again next time you log in.
