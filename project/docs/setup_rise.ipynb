# RISE slideshow setup

Run this notebook once to install the RISE slideshow tool. It installs into your home folder so it stays installed every time you log in, and sets Alt+R to open a notebook as slides.

Steps:

1. Run all the cells (Run menu, then Run All Cells).
2. Check that the output says everything is set.
3. Restart your server using the instructions in the last cell. A browser refresh does not
   do this.

After that, open soci_280_bert.ipynb or text_embeddings_workshop.ipynb and press Alt+R.

```python
import sys, subprocess

# Install RISE into ~/.local so it stays installed after a server restart.
subprocess.run([sys.executable, "-m", "pip", "install", "--user",
                "--force-reinstall", "--no-deps", "jupyterlab_rise==0.43.1"], check=True)

r = subprocess.run([sys.executable, "-m", "jupyter", "server", "extension",
                    "enable", "jupyterlab_rise", "--user"],
                   capture_output=True, text=True)
print(r.stdout, r.stderr)
print("Done. Run the next two cells, then restart your server.")
```

```python
import json, os

# Make Alt+R open the slides in a new tab. The default Alt+R opens the in-page preview, which does not work on JupyterOpen.
d = os.path.expanduser("~/.jupyter/lab/user-settings/@jupyterlab/shortcuts-extension")
os.makedirs(d, exist_ok=True)
shortcuts = {
    "shortcuts": [
        {"command": "RISE:preview", "keys": ["Alt R"],
         "selector": ".jp-Notebook:focus", "disabled": True},
        {"command": "RISE:slideshow", "keys": ["Alt R"],
         "selector": ".jp-Notebook:focus"}
    ]
}
with open(os.path.join(d, "shortcuts.jupyterlab-settings"), "w", encoding="utf-8") as f:
    json.dump(shortcuts, f, indent=2)
print("Alt+R will open the slides in a new tab after you restart.")
```

```python
import site, os

ul = site.getusersitepackages()
sc = os.path.expanduser(
    "~/.jupyter/lab/user-settings/@jupyterlab/shortcuts-extension/shortcuts.jupyterlab-settings")

print("user site enabled:", site.ENABLE_USER_SITE)
print("rise installed in home:", os.path.isdir(os.path.join(ul, "jupyterlab_rise")))
print("Alt+R shortcut written:", os.path.isfile(sc))
print("If all three say True, restart your server (see the last cell).")
```

## Restart your server

The server has to restart before it picks up RISE. A browser refresh will not work.

1. In the menu bar, go to File, then Hub Control Panel. You can also open
   https://open.jupyter.ubc.ca/jupyter/hub/home in a new tab.
2. Click Stop My Server and wait for it to stop.
3. Click Start My Server.
4. When JupyterLab loads again, open a notebook like docs/SOCI-280/soci_280_bert.ipynb and
   press Alt+R. The slides open in a new tab.

You do not need to run this setup again next time you log in.
